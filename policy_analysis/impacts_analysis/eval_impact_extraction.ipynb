```python
rename_dict = {
        'Resource': 'Natural Resources',
        'Wellbeing': 'Wellbeing',
        'Need': 'Human Needs',
        'PlanetaryBoundary': 'Planetary Boundaries',
        'Justice': 'Justice',
}
gold['impact_category'] = gold['impact_category'].map(rename_dict)
gold
```

```python
gold["impact_dim"] = gold.apply(
    lambda row: f"{row['impact_category']} - {row['impact'].replace('_', ' ').capitalize()}",
    axis=1,
)
```

```python
gold = gold.drop(columns=['impact_category', 'impact'])
gold
```

```python
with open("IMPACT_DIRECTION_EXTRACTION_PROMPT.txt") as f:
    system_prompt = f.read().strip()
```

```python
all_prompts = []
all_metadata = []
for row in gold.itertuples():
    prompt = f"{system_prompt}\n\nReference text:\n{row.text}\n\nPolicy: {row.policy}\nImpact Dimension: {row.impact_dim}\nImpact direction:"
    all_prompts.append(prompt)
    all_metadata.append({
        'openalex_id': row.openalex_id,
        'chunk_idx': int(row.chunk_idx),
        'row_id': int(row.Index),
        'impact_dim': row.impact_dim,
        'label': row.direction,
    })
```

```python
import os
import json
from dotenv import load_dotenv
from openai import AsyncOpenAI, RateLimitError
import asyncio
load_dotenv()
```

```python
client = AsyncOpenAI(
    base_url=os.getenv("GENERATION_API_URL"),
    api_key=os.getenv("SCW_SECRET_KEY"),
)
```

```python
choices = ['positive', 'negative', 'neutral', 'unknown']
model = "mistral-small-3.2-24b-instruct-2506"
CONCURRENCY = 20

response_format = {
    "type": "json_schema",
    "json_schema": {
        "name": "impact_direction",
        "strict": True,
        "schema": {
            "type": "object",
            "properties": {
                "impact_direction": {"type": "string", "enum": choices}
            },
            "required": ["impact_direction"],
            "additionalProperties": False,
        },
    },
}

sem = asyncio.Semaphore(CONCURRENCY)


async def predict_one(prompt, meta, max_retries=8):
    delay = 5.0
    for attempt in range(max_retries):
        async with sem:
            try:
                response = await client.chat.completions.create(
                    model=model,
                    messages=[{"role": "user", "content": prompt}],
                    max_tokens=10,
                    temperature=0,
                    response_format=response_format,
                )
                raw = response.choices[0].message.content
                label = json.loads(raw)["impact_direction"]
                return meta | {"impact_direction": label}
            except RateLimitError:
                if attempt == max_retries - 1:
                    raise
                await asyncio.sleep(delay)
                delay *= 2
```

```python
from tqdm.asyncio import tqdm_asyncio

async def predict_all():
    tasks = [predict_one(p, m) for p, m in zip(all_prompts, all_metadata)]
    return await tqdm_asyncio.gather(*tasks)

results = await predict_all()
```

```python
res = pd.DataFrame(results)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(res['label'], res['impact_direction'], zero_division=0))
```
